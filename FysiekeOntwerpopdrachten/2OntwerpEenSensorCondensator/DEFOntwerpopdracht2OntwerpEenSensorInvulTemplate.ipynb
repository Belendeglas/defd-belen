```python
# Schrijf hier een python functie die de capaciteit van een plaatcondensator uitrekent. Zorg dat de functie
# als input de ontwerpparameters van de vlakke plaat formule accepteert (oppervlakte, afstand en diëlektrische
# constante) en als output de capaciteit van de vlakke plaat condensator met die ontwerpparameters.
# Je gaat deze formule de rest van de dag gebruiken: controleer hem goed!
import numpy as np

epsilon_r_0= 2.6
A_0=0.01
d_0=0.0021
epsilon_0 = 8.854e-12
epsilon = epsilon_0 * epsilon_r_0

def plaat(A, d):
    return epsilon * A / d

print(plaat(A_0,d_0))
C=plaat(A_0,d_0)
R=1000
f=100000
def Vout(C):
    return 16*np.pi*f*R*C
V_out= Vout(C)
print(V_out)

#formule                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                
def vlakkeplaat(epsilon_r, A, d):
    epsilon_0 = 8.854e-12
    
    epsilon = epsilon_0 * epsilon_r
    
    C = (epsilon * A / d)*1e12
    
    return C


# Optioneel: schrijf je functie zodanig dat je arrays als input kan geven en (dus) ook arrays als output.
```

```python
# Gebruik je formule om uit te rekenen wat je verwacht dat de capaciteit van jullie vlakke plaat condensatoren
# gaat zijn.



C = [vlakkeplaat(epsilon_r_0,A_0,d_0), vlakkeplaat(1.3,A_0,0.018),vlakkeplaat(1.3,A_0,0.014),vlakkeplaat(1.3,A_0,0.020)]


print(f"Berekende capaciteit plaatcondensator groep 61 is {C[0]:.1f} pF")
```

## Opdracht 4: Uitgewerkte schets (afmetingen!) en plot van de ontwerpgrafiek.

#### Voeg hier een foto van jullie uitgewerkte  schets van de sensor die jullie gaan maken in
![uitgewerkte schets](schets_nieuw.jpeg)

```python
# Plot hier de verwachte ontwerpgrafiek van jullie sensoren.
#We gaan uit van een linear varband tot 1 cm dikte
#We plotten de massa op de x-as en de verwachte dikte op de y-as
import matplotlib.pyplot as plt 
from scipy.optimize import curve_fit
d_ontwerp=np.array([0.096,0.192,0.384,0.96])*1e-3
C_ontwerp = [vlakkeplaat(epsilon_r_0,A_0,d_ontwerp[0]), \
             vlakkeplaat(epsilon_r_0,A_0,d_ontwerp[1]),\
                vlakkeplaat(epsilon_r_0,A_0,d_ontwerp[2]),\
                    vlakkeplaat(epsilon_r_0,A_0,d_ontwerp[3])]
A4_ontwerp= d_ontwerp/0.000096

def functie(A4_ontwerp,a):
    return a/A4_ontwerp

val,cov = curve_fit(functie,A4_ontwerp,C_ontwerp)
A4_test= np.linspace(0,30,30)

plt.figure()
plt.title('Ontwerpgrafiek')
plt.xlabel('Aantal A4tjes ')
plt.ylabel('Capaciteit [pF]')
plt.ylim()
plt.scatter(A4_ontwerp,C_ontwerp)
plt.plot(A4_test,functie(A4_test,val[0]))
plt.show()

```

## Opdracht 7: Plot de kalibratiegrafiek.

```python
# Plot hier je gemeten kalibratiegrafiek en plot in dezelfde figuur ook je verwachte ontwerpgrafiek.
#Metingen met de hand
Vpp= np.array([3.56,2.86,2.64,2.10,1.84,1.08,0.940,0.880])
Aantal_A4= np.array([2,3,4,6,8,10,12,14,])
#pas vanaf 2 A4 is de afstand groot genoeg
def C_metV(Vpp):
    return (Vpp/2)/(R*16*np.pi*f)*1e12

C_kali= C_metV(Vpp)

def A4_C(Aantal_A4,b):
    return b/Aantal_A4
val1,cov1=curve_fit(A4_C,Aantal_A4,C_kali)
u_C_kali= 1.5

plt.figure()
plt.errorbar(Aantal_A4, C_kali, yerr=u_C_kali, fmt='o', capsize=4, color='green', label='metingen met fout')
plt.xlabel('Aantal A4tjes ')
plt.ylabel('Capaciteit [pF]')
plt.scatter(A4_ontwerp,C_ontwerp)
plt.scatter(Aantal_A4,C_kali)
plt.plot(A4_test,functie(A4_test,val[0]),label='ontwerp')
plt.plot(A4_test,A4_C(A4_test,val1[0]),label='metingen')
plt.legend()
plt.show()

```

## Opdracht 10: Vul hieronder de conclusie in.

De ontwerpgrafiek ligt veel hoger dan de grafieken van de uiteindelijke metingen. Dit kont waarschijnlijk omdat de dikte van 1 A4 niet helemaal correct is.
